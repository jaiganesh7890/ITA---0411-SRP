Experiment 12: Data reshaping and monthly averages using airquality

```r
# Load built-in dataset
data(airquality)

# Display original data
print(head(airquality))

# Monthly averages
monthly_avg <- aggregate(
  cbind(Ozone, Solar.R, Wind, Temp) ~ Month,
  data=airquality,
  FUN=function(x) mean(x, na.rm=TRUE)
)

print(monthly_avg)

# Reshape data into long format
long_data <- reshape(
  monthly_avg,
  varying=c("Ozone", "Solar.R", "Wind", "Temp"),
  v.names="Average",
  timevar="Variable",
  times=c("Ozone", "Solar.R", "Wind", "Temp"),
  direction="long"
)

print(long_data)
```

Experiment 15: Exploratory data analysis on the iris dataset

```r
data(iris)

# Dataset dimensions
print(dim(iris))

# Summary statistics
print(summary(iris))

# Standard deviation of numeric columns
print(sapply(iris[1:4], sd))

# Quantiles
print(sapply(iris[1:4], quantile))

# Grouping by Species
print(aggregate(. ~ Species, data=iris, FUN=mean))

# Pivot-style summary: mean Sepal.Length
pivot_table <- aggregate(
  Sepal.Length ~ Species,
  data=iris,
  FUN=mean
)
print(pivot_table)

# Categorize Sepal.Length
iris$SepalCategory <- cut(
  iris$Sepal.Length,
  breaks=3,
  labels=c("Short", "Medium", "Long")
)

print(table(iris$Species, iris$SepalCategory))
```

Experiment 16: Explore the USArrests dataset

```r
data(USArrests)

# Summary statistics
print(summary(USArrests))

# State with the largest rape arrest rate
print(rownames(USArrests)[which.max(USArrests$Rape)])

# Maximum and minimum murder rates
print(max(USArrests$Murder))
print(min(USArrests$Murder))

# Correlation among variables
print(cor(USArrests))

# States above median Assault
print(USArrests[USArrests$Assault > median(USArrests$Assault), ])

# Bottom 25% of Murder rates
cutoff <- quantile(USArrests$Murder, 0.25)
print(USArrests[USArrests$Murder <= cutoff, ])

# Histogram
hist(USArrests$Murder, main="Murder Arrest Rates",
     xlab="Murder", col="lightblue")

# Density plot
plot(density(USArrests$Murder),
     main="Density of Murder Arrest Rates")

# Scatterplot
plot(USArrests$Assault, USArrests$Murder,
     xlab="Assault", ylab="Murder",
     main="Assault vs Murder")

# Bar graph for the 10 states with highest Murder rates
top10 <- head(USArrests[order(-USArrests$Murder), ], 10)
barplot(top10$Murder, names.arg=rownames(top10),
        las=2, col="lightgreen",
        main="Top 10 States by Murder Rate")
```

Experiment 17: Explore the Titanic dataset

```r
data(Titanic)

# Convert the built-in Titanic table to a data frame
titanic_data <- as.data.frame(Titanic)

# Survival by passenger class
survival_class <- xtabs(Freq ~ Class + Survived,
                        data=titanic_data)
barplot(t(survival_class), beside=TRUE,
        col=c("tomato", "lightgreen"),
        legend=TRUE,
        main="Titanic Survival by Class",
        xlab="Passenger Class",
        ylab="Number of Passengers")

# Survival by gender
survival_gender <- xtabs(Freq ~ Sex + Survived,
                         data=titanic_data)
barplot(t(survival_gender), beside=TRUE,
        col=c("tomato", "lightgreen"),
        legend=TRUE,
        main="Titanic Survival by Gender",
        xlab="Gender",
        ylab="Number of Passengers")
```

Experiment 19: Build a regression model using the Advertising dataset

```r
# Load the dataset directly from a public URL to guarantee availability
url <- "https://www.statlearning.com/s/Advertising.csv"
Advertising_data <- read.csv(url)

# Display dataset
print(head(Advertising_data))

# Build regression model
model <- lm(sales ~ TV, data=Advertising_data)

# Display model results
print(summary(model))

# Predict Sales
predicted_sales <- predict(model, newdata=Advertising_data)
print(head(predicted_sales))

# Plot actual and predicted values
plot(Advertising_data$TV, Advertising_data$sales,
     main="Advertising Spend vs Sales",
     xlab="TV Advertising Spend",
     ylab="Sales",
     pch=19)

abline(model, col="red", lwd=2)
```

Experiment 20: Multiple regression using ChickWeight

```r
data(ChickWeight)

# Build multiple regression model
model <- lm(weight ~ Time + Diet, data=ChickWeight)

# Display model summary
print(summary(model))

# Predict weight
predicted_weight <- predict(model, newdata=ChickWeight)
print(head(predicted_weight))

# Calculate model error
actual_weight <- ChickWeight$weight

MAE <- mean(abs(actual_weight - predicted_weight))
MSE <- mean((actual_weight - predicted_weight)^2)
RMSE <- sqrt(MSE)

cat("MAE =", MAE, "\n")
cat("MSE =", MSE, "\n")
cat("RMSE =", RMSE, "\n")

# Compare actual and predicted weights
plot(actual_weight, predicted_weight,
     xlab="Actual Weight",
     ylab="Predicted Weight",
     main="Actual vs Predicted Weight",
     pch=19, col="blue")

abline(0, 1, col="red", lwd=2)
```

Experiment 21: Logistic regression using the iris dataset

```r
data(iris)

# Create binary target variable
iris$Target <- ifelse(iris$Species == "setosa", 1, 0)

# Randomly split data into training and testing sets
set.seed(123)

train_index <- sample(
  seq_len(nrow(iris)),
  size=0.8 * nrow(iris)
)

train_data <- iris[train_index, ]
test_data <- iris[-train_index, ]

# Build logistic regression model
model <- glm(
  Target ~ Petal.Length + Petal.Width,
  data=train_data,
  family=binomial
)

# Display model summary
print(summary(model))

# Predict probabilities
probabilities <- predict(model, newdata=test_data,
                         type="response")

# Convert probabilities into classes
predicted <- ifelse(probabilities >= 0.5, 1, 0)
actual <- test_data$Target

# Confusion matrix
confusion_matrix <- table(
  Actual=factor(actual, levels=c(0, 1)),
  Predicted=factor(predicted, levels=c(0, 1))
)

print(confusion_matrix)

# Calculate accuracy
accuracy <- mean(predicted == actual) * 100
cat("Accuracy:", round(accuracy, 2), "%\n")
```
