### 1. Check if a Number is Positive, Negative, or Zero in R

```r
check_positive_negative <- function(num) {
  if (num > 0) {
    return("Positive")
  } else if (num < 0) {
    return("Negative")
  } else {
    return("Zero")
  }
}

# Test the function
test_num <- -15
cat(paste("The number", test_num, "is", check_positive_negative(test_num), "\n"))
```

### 2. Sequence from Start and in Reverse Order in R

```r
start_val <- 1
end_val <- 10

cat("Sequence in starting order:\n")
cat(seq(start_val, end_val), "\n\n")

cat("Sequence in reverse order:\n")
cat(seq(end_val, start_val), "\n")
```

### 3. Check if a Number is Odd or Even in R

```r
check_odd_even <- function(num) {
  if (num %% 2 == 0) {
    return("Even")
  } else {
    return("Odd")
  }
}

# Test the function
test_val <- 7
cat(paste("The number", test_val, "is", check_odd_even(test_val), "\n"))
```

### 4. Check if a Number is an Armstrong Number in R

```r
is_armstrong <- function(num) {
  digits <- as.numeric(strsplit(as.character(num), "")[[1]])
  n <- length(digits)
  total_sum <- sum(digits ^ n)
  return(total_sum == num)
}

# Test the function
test_armstrong <- 153
if (is_armstrong(test_armstrong)) {
  cat(test_armstrong, "is an Armstrong number.\n")
} else {
  cat(test_armstrong, "is not an Armstrong number.\n")
}
```

### 5. Reverse the Digits of a Given Number in R

```r
reverse_digits <- function(num) {
  sign_val <- sign(num)
  abs_num <- abs(num)

  # Convert to characters, split, reverse, paste, and cast back to numeric
  reversed_chars <- rev(strsplit(as.character(abs_num), "")[[1]])
  reversed_num <- as.numeric(paste(reversed_chars, collapse = ""))

  return(sign_val * reversed_num)
}

# Test the function
test_reverse <- -12345
cat("Original:", test_reverse, "-> Reversed:", reverse_digits(test_reverse), "\n")
```
